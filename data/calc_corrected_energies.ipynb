```python
#!/usr/bin/env python

import numpy as np
import os
from ase.units import eV, Hartree

# cwd = os.getcwd()
structures_path = os.path.abspath('/Users/kb6562/KIT/2025/publication/EtOligo_Kinetics/Ni-II-NU-1000/Cossee-Arlman/data/structures')

models = ['QM-QM_M06L-PBED3', 'QM-QM_TPSSh-PBED3']
pathways = ['trigonal', 'hexagonal']

for model in models:
    for pathway in pathways:
        struc_path = os.path.join(structures_path, model, pathway)
        pbc_path = os.path.join(structures_path, 'pbc_PBED3BJ', pathway)
        pbe_path = os.path.join(structures_path, 'QM-QM_PBED3-PBED3', pathway)
        for dir in os.listdir(struc_path):
            dir_path = os.path.join(struc_path, dir)
            if not os.path.islink(dir_path):
                E_cluster = float(np.loadtxt(os.path.join(dir_path, 'energy_cluster.dat'))) * Hartree # Eh -> eV
                E_pbe = float(np.loadtxt(os.path.join(pbe_path, dir, 'energy_cluster.dat'))) * Hartree # Eh -> eV
                E_pbc = float(np.loadtxt(os.path.join(pbc_path, dir, 'energy.dat'))) # in eV
                E_corr = E_pbc + (E_cluster - E_pbe)
                with open(os.path.join(dir_path, 'energy.dat'), 'w') as fil:
                    fil.write(f'{E_corr}')
                # print(f"{model}\t{pathway}\t{dir}:\t\t{E_corr}")

molecules_path = os.path.abspath('/Users/kb6562/KIT/2025/publication/EtOligo_Kinetics/Ni-II-NU-1000/Cossee-Arlman/data/gas-phase-molecules')
for model in models:
    mol_path = os.path.join(molecules_path, model)
    pbc_path = os.path.join(molecules_path, 'pbc_PBED3BJ')
    pbe_path = os.path.join(molecules_path, 'QM-QM_PBED3-PBED3')
    for dir in os.listdir(mol_path):
        dir_path = os.path.join(mol_path, dir)
        E_cluster = float(np.loadtxt(os.path.join(dir_path, 'energy_cluster.dat'))) * Hartree # Eh -> eV
        E_pbe = float(np.loadtxt(os.path.join(pbe_path, dir, 'energy_cluster.dat'))) * Hartree # Eh -> eV
        E_pbc = float(np.loadtxt(os.path.join(pbc_path, dir, 'energy.dat'))) # in eV
        E_corr = E_pbc + (E_cluster - E_pbe)
        with open(os.path.join(dir_path, 'energy.dat'), 'w') as fil:
                    fil.write(f'{E_corr}')
```
